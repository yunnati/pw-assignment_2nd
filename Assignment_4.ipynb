Q.1)Explain the complete ML lifecycle. How is it different from traditional Software Development Lifecycle (SDLC)?
- Machine Learning (ML) Lifecycle
1. Problem Definition(Identify the business problem and define objectives.)
2. Data Collection
Gather data from databases, APIs or external sources.)
3. Data Preprocessing(Handling missing values, Removing duplicates)
4. Exploratory Data Analysis (EDA)(Visualize data using graphs and charts.)
5. Feature Engineering(Select important features.)
6. Model Selection(appropriate algorithm based on the problem.)
7. Model Training(Train the selected algorithm using training data.)
8. Model Evaluation(test data using suitable metrics.)
9. Model Deployment(Deploy the trained model to production)
11. Monitoring and Maintenance
- Difference Between ML Lifecycle and Traditional SDLC
   - The ML lifecycle emphasizes data preparation, model training, evaluation, deployment, monitoring, and retraining to maintain predictive performance over time. In contrast, the traditional SDLC focuses on designing, developing, testing, deploying, and maintaining software based on predefined requirements. The most significant difference is that ML systems require continuous adaptation to changing data, where as traditional software primarily evolves in response to changing business or user requirements.

Q.2. Why is translating business needs into a data science problem the most critical step? Give one example where poor problem framing leads to wasted resources.
- translating business needs into a data science problem the most critical step:
    - Aligns with business goals: Ensures the project solves the right problem.
    - Defines measurable objectives: Establishes clear success metrics (e.g., reducing costs, increasing sales, improving customer satisfaction).
    - Guides data collection: Identifies what data is needed and avoids collecting unnecessary data.
    - Helps select the right ML approach: Determines whether the task is classification, regression, clustering, recommendation, etc.
    - Saves time and cost: Prevents effort from being spent on building models that do not solve the business need.

- Example of Poor Problem Framing Leading to Wasted Resources

    - Business Goal: Reduce customer churn in a telecom company:
          predict which customers will make the most phone calls next month instead of identifying customers likely to leave the service.
    -   Result:
        - The team spends months collecting call records and building a highly accurate prediction model.
        - The model successfully predicts call volume but does not identify customers at risk of leaving.
        - The company cannot take targeted retention actions, so customer churn remains high.
        - Significant time, money, and computing resources are wasted because the model solves the wrong problem.

Q.5: Dataset contains: ● 20% missing values ● Duplicate rows ● Imbalanced class distribution Explain step-by-step how you would clean and prepare this dataset.

- Step 1: Explore the dataset : Check data types, Identify missing values, Count duplicates, Examine class distribution .
- Step 2: Handle missing values : Remove columns with excessive missing values , Fill numerical values using mean or median, Fill categorical values using mode, Use advanced imputation techniques if needed.
- Step 3: Remove duplicate rows : Identify duplicate records, Remove exact duplicates, Verify dataset consistency.
- Step 4: Handle imbalanced classes : Use techniques such as: Oversampling , Undersampling , Balanced sampling ,etc.
- Step 5: Encode categorical variables : One-Hot Encoding , Label Encoding
- Step 6: Scale numerical features : Standardization , Normalization
- Step 7: Detect outliers using : Z-score , Isolation Forest
- Step 8: Split dataset : Training Set (70%) , Validation Set (15%) , Test Set (15%)


Q.6: Explain: ● Tokenization ● Embeddings ● Context window Why are embeddings critical for GenAI systems?
- 1. Tokenization : Tokenization is the process of breaking text into smaller units called tokens.
Ex., Sentence: "Machine learning is amazing."
Tokens: ["Machine", "learning", "is", "amazing", "."]
- 2. Embeddings : Embeddings are dense numerical vectors that represent words, sentences, or documents while preserving semantic meaning.
Ex., King → [0.24, -0.51, 0.83, ...] ,  Queen → [0.25, -0.49, 0.81, ...]
Similar words have similar vectors.
- 3. Context Window
The context window is the maximum amount of text (tokens) an AI model can process in one interaction.
- Ex., If the context window is 8,000 tokens, the model can only consider the latest 8,000 tokens when generating a response.
- Why Are Embeddings Critical for GenAI? : Capture semantic meaning , Improve search and retrieval , Enable Retrieval-Augmented Generation (RAG), Help recommend similar documents , Support question answering over large knowledge bases.


Q.7: Explain: ● Train/Validation/Test split ● Cross-validation ● Data leakage Why is data leakage dangerous?
- 1. Train/Validation/Test Split
    -	Training Set: Used to train the model.
    -	Validation Set: Used for hyperparameter tuning.
    -	Test Set: Used for final performance evaluation.
- 2. Cross-Validation :
Cross-validation divides the dataset into multiple folds (commonly 5 or 10).
Each fold is used once as the validation set while the remaining folds are used for training.
Benefits: Better performance estimation , Reduces overfitting ,Makes efficient use of data .
- 3. Data Leakage :
Data leakage occurs when information from the test or future data is unintentionally used during model training.
Examples:
    -	Including future sales data while predicting future demand.
    -	Scaling the entire dataset before splitting it into train and test sets.
- Why Is Data Leakage Dangerous?
    -	Produces unrealistically high accuracy.
    -	Gives misleading evaluation results.
    -	Causes poor performance in real-world deployment.
    -	Leads to incorrect business decisions.


 Q.8:Explain and compare: ● Accuracy ● Precision ● Recall ● F1-score ● ROC-AUC Why is accuracy misleading in fraud detection?
 - Accuracy : Accuracy measures the overall correctness of a machine learning model. It is the ratio of correctly predicted observations (both positive and negative) to the total number of observations. Accuracy is easy to understand and works well when the dataset is balanced. However, it can be misleading when the classes are imbalanced.
 - Precision : Precision measures how many of the instances predicted as positive are actually positive. It focuses on the quality of positive predictions. A high precision value means the model produces very few false positives. Precision is especially important in situations where false alarms are costly, such as spam email detection or fraud detection.

- Recall : Recall, also known as Sensitivity or True Positive Rate, measures how many of the actual positive instances are correctly identified by the model.
A high recall indicates that the model misses very few positive cases. Recall is crucial in applications such as disease diagnosis or security systems, where failing to detect a positive case can have serious consequences.
- F1-Score : The F1-score combines Precision and Recall into a single metric by calculating their harmonic mean. the F1-score provides a balanced measure when both false positives and false negatives are important. It is particularly useful for evaluating models on imbalanced datasets because it considers both Precision and Recall rather than relying only on Accuracy.
- ROC-AUC : ROC (Receiver Operating Characteristic) is a curve that plots the True Positive Rate (Recall) against the False Positive Rate at different classification thresholds. The AUC (Area Under the Curve) measures the overall ability of the model to distinguish between positive and negative classes.

- Comparison : Accuracy evaluates the overall percentage of correct predictions but may give misleading results when the dataset is imbalanced. Precision measures the reliability of positive predictions by minimizing false positives, making it suitable when false alarms are costly. Recall measures the model's ability to identify all actual positive cases by minimizing false negatives, making it essential when missing positive cases is unacceptable. The F1-score balances Precision and Recall into a single value and is preferred when both types of errors are equally important, especially with imbalanced data. ROC-AUC differs from the other metrics because it evaluates the model's ability to separate classes across all possible decision thresholds rather than at a single threshold. It provides a comprehensive measure of classification performance and is widely used for comparing different models.

- Why Is Accuracy Misleading in Fraud Detection?
Example:
    -	Total transactions = 10,000
    -	Fraud cases = 100
    -	Legitimate transactions = 9,900
    If a model predicts every transaction as legitimate:
    -	Accuracy = 99%
    -	Recall for fraud = 0%
Although the accuracy is very high, the model fails to detect any fraud. Therefore, precision, recall, F1-score, and ROC-AUC are more appropriate metrics for fraud detection.


Q.9: After 6 months: ● Model accuracy drops. ● Customer behavior changes. Explain: 1. Data drift 2. Concept drift 3. Monitoring techniques 4. Retraining strategy
- 1. Data Drift : Data drift occurs when the distribution of input data changes over time.
Example: Customer spending habits change due to inflation.
- 2. Concept Drift : Concept drift occurs when the relationship between inputs and outputs changes.
Example: Fraudsters adopt new techniques, making previous fraud patterns ineffective.
- 3. Monitoring Techniques
    -	Track prediction accuracy.
    - Monitor precision and recall.
    -	Detect changes in feature distributions.
    -	Monitor data quality.
    -	Set alerts for performance degradation.
    -	Collect user feedback.
- 4. Retraining Strategy
    -	Continuously collect new data.
    -	Validate data quality.
    -	Retrain the model periodically or when performance drops below a threshold.
    -	Compare the new model with the current production model.
    -	Deploy the improved model.
    -	Continue monitoring after deployment.

- 1. Data Flow
   - Customer Transaction
   - Data Collection
   - Data Preprocessing
   - ML Fraud Detection Model
        -   Fraud Score
   - Customer Support Request
   - GenAI Email Response Model
   - Draft Email Response
- 2. Model Interaction
    -	The ML fraud detection model analyzes transactions and assigns a fraud risk score. 	High-risk transactions are flagged for review.
    -	The Generative AI model creates automated email responses for customer queries, such as fraud alerts or account updates.
    -	The fraud detection output can be used to personalize the generated email.
- 3. Human-in-the-Loop
    -	Fraud analysts review transactions flagged as high risk before action is taken.
    -	Customer support agents review or edit AI-generated emails for sensitive or complex cases before sending them.
    -	Feedback from analysts and agents is used to improve both models over time.
- 4. Monitoring Checkpoints
    -	Data Quality: Monitor missing values, duplicates, and anomalies.
    -	Fraud Model: Track accuracy, precision, recall, F1-score, false positives, and false negatives.
    -	GenAI Model: Monitor response quality, relevance, hallucinations, response time, and user satisfaction.
    -	System Health: Monitor latency, uptime, API failures, and resource usage.
    -	Drift Detection: Detect data drift and concept drift, triggering retraining when performance degrades.
This architecture combines predictive ML for fraud detection with Generative AI for automated customer communication, while ensuring human oversight and continuous monitoring for reliable production performance.
